```python
class Employee:
    def __init__ (self, salary):
        self._salary = salary 
      
    @property   
    def salary(self):
       return self._salary
    
employee = Employee(25000)
employee.salary
employee.salary = 30000
```

```python
class Employee:
    def __init__ (self, salary):
        self._salary = salary 
      
    @property   
    def salary(self):
       return self._salary
    
    @salary.setter
    def salary(self, new_salary):
       self._salary = new_salary 
       return self._salary
    
employee = Employee(25000)
employee.salary = 35000
employee.salary


```
